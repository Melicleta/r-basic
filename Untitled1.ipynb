```python
# ==========================================
# Autor: Melissa Fuenzalida
# ==========================================

# ==========================================
# CARGA DE DATOS Y EXPLORACIÓN
# ==========================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from google.colab import drive

# Librerías para trabajo en Jupyter Notebook
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')

drive.mount("/content/drive")

# Cargar el archivo CSV
df = pd.read_csv('/content/drive/MyDrive/Ingenieria Ciencia de Datos/I Bimestre /Matematicas/Codigos/Matematicas_CDPython/data/Dataset_boston_housing_Forma A.csv')

# Ver las primeras filas y la estructura del dataset
print("=== PRIMERAS FILAS ===")
print(df.head())

print("\n=== INFORMACIÓN DE COLUMNAS ===")
print(df.info())
df.describe().round(4).T
```

```python
# ==========================================
# SELECCIÓN Y MATRIZ DE DISEÑO
# ==========================================
# Selección de variable dependiente (Y) e independientes (X)
# Justificación:
# - RM: Cantidad de habitaciones (relación positiva con el precio)
# - LSTAT: Porcentaje de población de menor ingreso (relación negativa)
# - CRIM: Tasa de criminalidad per cápita (relación negativa)

y = df['MEDV'].values.reshape(-1, 1)
X_variables = df[['RM', 'LSTAT', 'CRIM']].values

# Agregar la columna de unos para el término de sesgo (beta_0)
filas = len(y)
unos = np.ones((filas, 1))
X = np.hstack((unos, X_variables))

# Mostrar dimensiones de las matrices
print("\n=== DIMENSIONES ===")
print("Dimensión de X (Matriz de diseño):", X.shape)
print("Dimensión de Y (Vector objetivo):", y.shape)
```

```python
# ==========================================
# SECCIÓN 2: REGRESIÓN LINEAL Y ECUACIÓN NORMAL
# ==========================================
# Fórmula de la Ecuación Normal: beta = (X^T * X)^(-1) * X^T * Y
X_transpuesta = X.T
beta_ols = np.linalg.inv(X_transpuesta @ X) @ X_transpuesta @ y

# Calcular predicciones y el Error Cuadrático Medio (MSE)
predicciones_ols = X @ beta_ols
mse_ols = np.mean((y - predicciones_ols) ** 2)

print("\n=== MODELO SIN REGULARIZACIÓN (OLS) ===")
print("Beta 0 (Intercepto):", beta_ols[0][0])
print("Beta 1 (RM):", beta_ols[1][0])
print("Beta 2 (LSTAT):", beta_ols[2][0])
print("Beta 3 (CRIM):", beta_ols[3][0])
print("MSE (Error Cuadrático Medio):", mse_ols)
```

```python
# ==========================================
# SECCIÓN 3: REGULARIZACIÓN RIDGE (L2)
# ==========================================
# Función para calcular beta con regularización Ridge:
# beta_ridge = (X^T * X + lambda * I)^(-1) * X^T * Y

def calcular_ridge(X, y, lambda_val):
    columnas = X.shape[1]
    matriz_I = np.eye(columnas)
    matriz_I[0, 0] = 0  # No penalizar el intercepto

    beta_ridge = np.linalg.inv(X.T @ X + lambda_val * matriz_I) @ X.T @ y
    return beta_ridge

# Calcular coeficientes para lambda = 0.1
beta_ridge_01 = calcular_ridge(X, y, 0.1)
pred_r01 = X @ beta_ridge_01
mse_r01 = np.mean((y - pred_r01) ** 2)

# Calcular coeficientes para lambda = 1.0
beta_ridge_10 = calcular_ridge(X, y, 1.0)
pred_r10 = X @ beta_ridge_10
mse_r10 = np.mean((y - pred_r10) ** 2)

# Imprimir resultados comparativos
print("\n=== MODELO RIDGE (lambda = 0.1) ===")
print("Betas:", beta_ridge_01.ravel())
print("MSE:", mse_r01)

print("\n=== MODELO RIDGE (lambda = 1.0) ===")
print("Betas:", beta_ridge_10.ravel())
print("MSE:", mse_r10)
```
